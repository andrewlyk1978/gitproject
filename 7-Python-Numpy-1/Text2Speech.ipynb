```python
import pyttsx3
engine = pyttsx3.init()
newVoiceRate = 120
engine.setProperty('rate',newVoiceRate)
```

```python
# Voice selection options
voices = engine.getProperty('voices')
for voice in voices:
    print(voice)
```

```python
engine.setProperty('voice', voices[1].id)
engine.say("I love Python Analytics.")
engine.runAndWait()
```

```python
engine.setProperty('voice', voices[-12].id)
engine.say("呢個係入境事務處嘅最後通知。你有一份Python功課未交，請盡快完成。再見！")
engine.runAndWait()
```
