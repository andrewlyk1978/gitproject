## 55. Element-wise logical operations on two arrays
<pre>Find the logical value of a and b
Given Input:
a = np.array([False, True, False, True])
b = np.array([False, False, True, True])

Desired Output:
a and b: [False False False  True]
a or b: [False  True  True  True]
a xor b: [False  True  True False]</pre>


    

```python
aAAa=np.array([False, True, False, True])
bBBb=np.array([False, False, True, True])

aAAa_bBBb=np.logical_and(aAAa,bBBb)
aAAaOobBBb=np.logical_or(aAAa,bBBb)
aAAaμμbBBb=np.logical_xor(aAAa,bBBb)

print(aAAa_bBBb)
print(aAAaOobBBb)
print(aAAaμμbBBb)
```
