```python
def _distance(x,y):
    z=(x**2+y**2)**0.5
    print (f"x=", x ,"y=", y, "Distance_Z=",z )
    return z
    

distance=np.frompyfunc(_distance,2,1)

x=np.array([3,6,9])
y=np.array([4,8,12])
print(type(distance),distance(x,y))
```

```python
@np.vectorize
def finDistance(x,y):
      z=(x**2+y**2)**0.5
      print (f"x=", x ,"y=", y, "Distance_Z=",z )
      return z

x=np.array([3,6,9])
y=np.array([4,8,12])
print(finDistance(x,y))
print(type(finDistance))
```
