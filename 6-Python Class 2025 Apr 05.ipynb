```python
class Staff : 
    #Initialize instance attribute 
    def __init__(self,name,sid,gender):
        self.name:str=name
        self.sid:str=sid
        self.gender:list=gender
    
    #Instance method 
    def verify(self):
        return f"{self.name} ID:{self.sid} Gender:{self.gender}"
    def open_email(self):
        return f"{self.name.lower()}@com.hk"


#Create Sales Manager object 
sales_manager=Staff('Alice','E12345','Female')

#Verify Sales Manager
print(sales_manager.verify())
print(sales_manager.open_email())

    
```

```python
def sum_num(*args):
    total_args=0
    for arg in args: 
        total_args=total_args+arg
        print("total_args:",total_args,"arg:", arg, "sum_num:", sum_num)
    return total_args

print("sum_num(1):",sum_num(1))
print("sum_num(1,2,3):",sum_num(1,2,3))
print("sum_num(1,2,3,2):", sum_num(1,2,3,2))
```
