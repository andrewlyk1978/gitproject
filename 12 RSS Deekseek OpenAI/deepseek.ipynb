```python
import os
from openai import OpenAI
from IPython.display import display, Markdown, Latex

DEEPSEEK_API = os.environ['DEEPSEEK_API_KEY']  # Replace your key with "os.environ['DEEPSEEK_API_KEY']"

client = OpenAI(api_key = DEEPSEEK_API, base_url="https://api.deepseek.com")

response = client.chat.completions.create(
    model="deepseek-chat",
    messages=[
        {"role": "system", "content": "You are a helpful assistant and expert in quant finance"},
        {"role": "user", "content": "Hello"},
    ],
    stream=False
)

print(response.choices[0].message.content)
```

```python
question = 'What is smart beta? Output in Japanese in 100 words.'

messages = [{"role": "user", "content": question }]
response = client.chat.completions.create(
    model="deepseek-chat",
    messages=messages
)

messages.append(response.choices[0].message)
display(Markdown(response.choices[0].message.content))
```
