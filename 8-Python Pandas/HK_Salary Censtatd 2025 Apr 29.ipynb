Table 220-19013 : Average monthly salaries, average number of normal working hours per day and average number of standard working days per month of supervisory, technical, clerical, service and miscellaneous non-production workers by industry by occupation by sex [English]
https://www.censtatd.gov.hk/en/web_table.html?id=220-19013

API: https://www.censtatd.gov.hk/api/get.php?id=220-19013&lang=en&full_series=1

```python
import json
import requests
import pandas as pd

url='https://www.censtatd.gov.hk/api/get.php?id=220-19013&lang=en&full_series=1'
    
try:
    response = requests.get(url)
    if response.status_code==200 : print('Connect successfully!')
    data = response.json()
except Exception as e:
    print(e)    
```

```python
salary = pd.DataFrame(data['dataSet'])
salary.head(5)
```
